```python
import pandas as pd
# pandas module for cleaning dataset

df = pd.read_csv("C:\\Users\\Admin\\Documents\\Excel_all_Files\\SalesProject.csv",encoding = "latin-1")
# encoding - file format as on server
df.head()
# just to check if the dataset is correct. displays first 5 rows 
```

```python
print(df.isnull().sum())
#isnull() - function which checks for null values in each column 
# sum() - display the count of null values
```

```python
df.info()
# info() - give all the information about table
```

```python
# # changing datatype of date field
# df['Order Date'] = pd.to_datetime(df['Order Date'])
# df['Ship Date'] = pd.to_datetime(df['Ship Date'])
# df.info()
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.info()
```

```python
# # removing outliers - too small or too large to be removed
df = df[(df['Sales'] >= 25) & (df['Sales'] < 100000)]
df.info()
```

```python
# create seperate tables/datasets/dataframes from  single datasets
customer_df = df[['Customer ID','Customer Name','Segment','Country','City','State','Postal Code','Region']]
customer_df = customer_df.drop_duplicates(subset = 'Customer ID') 
customer_df.info()
```

```python

products_df = df[['Product ID', 'Category', 'Sub-Category', 'Product Name' ]] 
products_df = products_df.drop_duplicates(subset = 'Product ID') 
products_df.info()
```

```python
orders_df = df[['Row ID','Order ID','Order Date','Ship Date','Ship Mode','Customer ID','Product ID','Sales','Quantity','Discount','Profit']]
orders_df.info()
```

```python
# save/ export/load all three dataframes to excel files
with pd.ExcelWriter("C:\\Users\\Admin\\Documents\\Excel_all_Files\\SalesProject1.xlsx",engine ='xlsxwriter') as writer:
    customer_df.to_excel(writer, sheet_name = 'customer', index = False)
    product_df.to_excel(writer, sheet_name = 'product', index = False)
    orders_df.to_excel(writer, sheet_name = 'orders', index = False)
    print("File created")
```
